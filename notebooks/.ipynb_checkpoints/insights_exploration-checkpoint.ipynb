# Insights — exploration scratchpad

Goal: eyeball the data before deciding which event hypotheses survive into the Insights tab.

Hypotheses to look at:
- COVID lockdown (spring 2020) — drop in overall volume, shift toward residential
- George Floyd unrest (late May / early June 2020) — short sharp spike, then settle
- Laquan McDonald video release (Nov 2015) — arrest rate hypothesis, not volume
- Polar vortex / blizzards — short troughs in winter daily volume
- Election windows under incumbent mayors — week-of-vote rhythm

Source: the local `chicago_crimes.csv` mirror (1.9 GB, gitignored). Going through pandas is faster and more interactive than paginating Socrata for 8.5M rows. The Socrata script in `scripts/fetch_and_build.py` remains the production source — this notebook is just a scratchpad.

**All event dates in this notebook are placeholders. Verify before publishing anything.**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

plt.rcParams['figure.figsize'] = (12, 4)
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False

CSV_PATH = Path('..') / 'chicago_crimes.csv'
CSV_PATH.resolve()
```

## Load

Only the columns we need for these questions. `Date` parsed up-front, dtypes minimised.

```python
usecols = ['Date', 'Primary Type', 'Arrest', 'Community Area', 'Location Description']
dtype = {
    'Primary Type': 'category',
    'Location Description': 'category',
    'Arrest': 'bool',
    'Community Area': 'float32',
}

df = pd.read_csv(
    CSV_PATH,
    usecols=usecols,
    dtype=dtype,
    parse_dates=['Date'],
    date_format='%m/%d/%Y %I:%M:%S %p',
)
df = df.rename(columns={
    'Date': 'date',
    'Primary Type': 'type',
    'Arrest': 'arrest',
    'Community Area': 'ca',
    'Location Description': 'location',
})
df = df.sort_values('date').reset_index(drop=True)
df['day'] = df['date'].dt.normalize()
print(f'{len(df):,} rows')
print(f'range: {df.date.min()}  →  {df.date.max()}')
df.head()
```

## 1. Long trend — daily volume, 2001 → today

Look for: COVID dip, secular decline, any sharp ceiling/floor moves.

```python
daily = df.groupby('day').size().rename('n').to_frame()
daily['n_7d'] = daily['n'].rolling(7, center=True).mean()
daily['n_28d'] = daily['n'].rolling(28, center=True).mean()

fig, ax = plt.subplots()
ax.plot(daily.index, daily['n'], lw=0.4, color='#c93023', alpha=0.45, label='Daily')
ax.plot(daily.index, daily['n_28d'], lw=1.4, color='#5a0a0a', label='28-day rolling')
ax.set_title('Reported incidents per day · 2001 → today')
ax.set_ylabel('Incidents / day')
ax.legend()
plt.tight_layout()
```

## 2. Spring 2020 — COVID lockdown + Floyd protests

Candidate event markers (**verify before quoting**):
- ~2020-03-21 — Illinois "stay-at-home" order took effect
- ~2020-05-25 — George Floyd killed in Minneapolis
- ~2020-05-30 → 2020-06-07 — Chicago unrest peak

Looking for: cliff-edge drop in mid-March, then a sharp transient bump late May / early June.

```python
win = daily.loc['2020-01-01':'2020-09-30']
fig, ax = plt.subplots()
ax.plot(win.index, win['n'], lw=0.7, color='#c93023', alpha=0.6, label='Daily')
ax.plot(win.index, win['n_7d'], lw=1.6, color='#5a0a0a', label='7-day rolling')
for date, label in [('2020-03-21', 'IL stay-at-home (verify)'),
                    ('2020-05-25', 'Floyd killed (verify)'),
                    ('2020-05-30', 'Chicago unrest (verify)')]:
    ax.axvline(pd.Timestamp(date), color='#1a1a1a', lw=0.7, ls='--', alpha=0.6)
    ax.text(pd.Timestamp(date), ax.get_ylim()[1]*0.95, '  '+label, fontsize=8, rotation=90, va='top')
ax.set_title('Spring–summer 2020 · daily volume')
ax.legend()
plt.tight_layout()
```

## 3. November 2015 — Laquan McDonald video release

Candidate event marker (**verify**): ~2015-11-24, dashcam footage released by court order.

Hypothesis: arrest **rate** drops more visibly than volume — "depolicing" effect.

```python
win = daily.loc['2015-09-01':'2016-03-31']
fig, ax = plt.subplots()
ax.plot(win.index, win['n'], lw=0.7, color='#c93023', alpha=0.6, label='Daily')
ax.plot(win.index, win['n_7d'], lw=1.6, color='#5a0a0a', label='7-day rolling')
ax.axvline(pd.Timestamp('2015-11-24'), color='#1a1a1a', lw=0.7, ls='--')
ax.text(pd.Timestamp('2015-11-24'), ax.get_ylim()[1]*0.95, '  Laquan video (verify)', fontsize=8, rotation=90, va='top')
ax.set_title('Late 2015 → early 2016 · daily volume')
ax.legend()
plt.tight_layout()
```

## 4. Blizzards / polar vortex candidates

Placeholder dates (**verify, none are confirmed yet**):
- 2011-02-01 → 2011-02-03 — "Groundhog Day blizzard"
- 2014-01-06 → 2014-01-08 — polar vortex peak
- 2019-01-30 → 2019-02-01 — Chicago polar vortex

Pick a window per event and look for a 1-3 day trough in volume.

```python
for start, end, label in [
    ('2011-01-15', '2011-02-20', '2011 Groundhog Day blizzard (verify)'),
    ('2013-12-20', '2014-01-31', '2014 polar vortex (verify)'),
    ('2019-01-15', '2019-02-15', '2019 polar vortex (verify)'),
]:
    win = daily.loc[start:end]
    fig, ax = plt.subplots(figsize=(10, 2.8))
    ax.plot(win.index, win['n'], lw=1, color='#c93023', alpha=0.85)
    ax.set_title(label)
    ax.set_ylabel('Incidents / day')
    plt.tight_layout()
    plt.show()
```

## 5. Arrest rate over time

Monthly share of incidents with `arrest=True`. The Laquan hypothesis lives here, not in volume.

```python
monthly = df.set_index('date').resample('MS').agg(
    n=('arrest', 'size'),
    arrests=('arrest', 'sum'),
)
monthly['arrest_rate'] = monthly['arrests'] / monthly['n']

fig, ax = plt.subplots()
ax.plot(monthly.index, monthly['arrest_rate'], lw=1.2, color='#5a0a0a')
ax.axhline(0.251, color='#1a1a1a', lw=0.6, ls='--', label='25.1% baseline')
ax.axvline(pd.Timestamp('2015-11-24'), color='#c93023', lw=0.7, ls='--', alpha=0.7)
ax.text(pd.Timestamp('2015-11-24'), 0.45, '  Laquan video (verify)', fontsize=8, rotation=90, va='top')
ax.set_title('Monthly arrest rate · 2001 → today')
ax.set_ylabel('Share of incidents with arrest')
ax.legend()
plt.tight_layout()
```

## 6. Breakdown by primary type

Which types carry the bulk of volume, and which respond visibly to specific events?

```python
type_totals = df['type'].value_counts()
top = type_totals.head(8).index.tolist()
print('top 8:', top)
print()
print(type_totals.head(15))
```

```python
monthly_by_type = (
    df[df['type'].isin(top)]
    .set_index('date')
    .groupby('type', observed=True)
    .resample('MS')
    .size()
    .unstack(level=0)
    .fillna(0)
)

fig, ax = plt.subplots(figsize=(12, 5))
for t in monthly_by_type.columns:
    ax.plot(monthly_by_type.index, monthly_by_type[t], lw=1.1, label=t.title())
ax.set_title('Monthly volume by primary type · top 8')
ax.legend(ncol=2, fontsize=8, loc='upper right')
plt.tight_layout()
```

## 7. Election windows under incumbent mayors

Placeholder dates (**verify**):
- 2008-11-04 — Obama elected (Daley mayor)
- 2012-11-06 — Obama reelected (Emanuel mayor)
- 2016-11-08 — Trump elected (Emanuel mayor)
- 2020-11-03 — Biden elected (Lightfoot mayor)
- 2024-11-05 — placeholder (verify date and mayor)

Plot ±21 days around each, indexed to the centred 28-day rolling mean.

```python
elections = ['2008-11-04', '2012-11-06', '2016-11-08', '2020-11-03', '2024-11-05']
WINDOW = 21

fig, ax = plt.subplots()
for d in elections:
    centre = pd.Timestamp(d)
    win = daily.loc[centre - pd.Timedelta(days=WINDOW): centre + pd.Timedelta(days=WINDOW)].copy()
    baseline = win['n'].mean()
    win['idx'] = (win.index - centre).days
    win['ratio'] = win['n'] / baseline
    ax.plot(win['idx'], win['ratio'], lw=1, alpha=0.85, label=d[:4])
ax.axvline(0, color='#1a1a1a', lw=0.6, ls='--')
ax.axhline(1, color='#888', lw=0.6, ls=':')
ax.set_title('Election windows · daily volume / window mean (election day = 0)')
ax.set_xlabel('Days from election day')
ax.legend(fontsize=9)
plt.tight_layout()
```
